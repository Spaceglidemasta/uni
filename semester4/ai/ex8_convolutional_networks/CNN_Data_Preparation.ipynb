# Cats vs Dogs Dataset

The Cats vs Dogs dataset from Kaggle is a classic image classification benchmark used for training 
and evaluating computer vision models. Originally hosted on Kaggle as part of a competition in 2013.

Key Characteristics:
-Binary classification problem
-Reflects real-world challenges 
 i.e. varied poses and lighting conditions,occlusion and background clutter, different breeds and appearances
 
Common Uses: 
-Benchmarking CNNs (Convolutional Neural Networks)
-Learning image preprocessing, augmentation, transfer learning
-Practicing data pipeline construction (train/val/test split)

The dataset consist of
Training set: ~25,000 labeled images (filename indicates label)
Test set: ~12,500 unlabeled images (used for competition submission)

Image format: JPEG (.jpg)
Image size: Varies (not fixed resolution)
Structure (after extraction)

```python
import os
from os import makedirs
from os import listdir
from shutil import copyfile
from random import seed
from random import random


# # Directory of the original dataset (relative to your homedir) - DO NOT CHANGE
## You may have to adapt the relative path depending on your notebooks directory 


```

## Get ready - Data Preparation

Split the original training set from the archive in training and validation data

```python
SRC_DIR = 'PetImages/'
DOG_TRAIN = "data-dogs-vs-cats/train/dogs/"
CAT_TRAIN  = "data-dogs-vs-cats/train/cats/"
DOG_VAL = "data-dogs-vs-cats/validation/dogs/"
CAT_VAL  = "data-dogs-vs-cats/validation/cats/"


# #  Home Directory for Learning
dataset_home='data-dogs-vs-cats/'

# # create new sudirectories 
subdirs = ['train/', 'validation/']
for subdir in subdirs:

    # create label subdirectories
     labeldirs = ['dogs/', 'cats/']
     for labldir in labeldirs:
         newdir = dataset_home + subdir + labldir
         makedirs(newdir, exist_ok=True)

# # seed random number generator
seed(42)  


# # define ratio of pictures to use for validation
val_ratio = 0.25


# # Split the data into training and validation set
for file in listdir(SRC_DIR + "Cat"):
    
    src = SRC_DIR + 'Cat/' + file
    
    if random() < val_ratio:

        dst = CAT_TRAIN + file
        copyfile(src, dst)

    else:

        dst = CAT_VAL + file
        copyfile(src, dst)


for file in listdir(SRC_DIR + "Dog"):
    
    src = SRC_DIR + 'Dog/' + file
    
    if random() < val_ratio:
        
        dst = DOG_TRAIN  + file
        copyfile(src, dst)

    else:
        
        dst = DOG_VAL + file
        copyfile(src, dst)



# # check statistics of the split
print("Training Data:")
files = [
    f for f in listdir(dataset_home + 'train/' + 'cats/')
    if os.path.isfile(os.path.join(SRC_DIR, f)) and not f.startswith('.')
]
num_files = len(files)
print(f"Number of cat images  {num_files}")
files = [
    f for f in listdir(dataset_home + 'train/' + 'dogs/')
    if os.path.isfile(os.path.join(SRC_DIR, f)) and not f.startswith('.')
]
num_files = len(files)
print(f"Number of dog images  {num_files}")

print("Validation Data:")
files = [
    f for f in listdir(dataset_home + 'validation/' + 'cats/')
    if os.path.isfile(os.path.join(SRC_DIR, f)) and not f.startswith('.')
]
num_files = len(files)
print(f"Number of cat images  {num_files}")
files = [
    f for f in listdir(dataset_home + 'validation/' + 'dogs/')
    if os.path.isfile(os.path.join(SRC_DIR, f)) and not f.startswith('.')
]
num_files = len(files)
print(f"Number of dog images  {num_files}")
```
