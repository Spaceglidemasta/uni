```python
# baseline model for the dogs vs cats dataset



import matplotlib
from matplotlib import pyplot

from keras.models import Sequential
from keras.layers import Input
from keras.layers import Conv2D
from keras.layers import MaxPooling2D
from keras.layers import Dense
from keras.layers import Flatten

from keras.optimizers import SGD
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# from keras.models import load_model
```

# Let's start with setting up the CNN

```python
# method defining a simple CNN
# first use it as specified, in a second and third version add Task 1 and Task 2
#
def define_model():
    # defines the CNN model architecture
    model = Sequential()
    model.add(Input(shape=(200, 200, 3)))  # Explicit input layer
    # Convolution Layer
    model.add(Conv2D(32, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'))
    model.add(MaxPooling2D((2, 2)))
    
## Task 1 (after training with one convolutional layer):     
# Add a second convolutional layer with 64 filter kernel (3x3), don't forget the pooling layer
# Stick to rest of the settings!
#



## Task 2 (after training with two convolutional layer):     
# Add a third convolutional layer with 128 filter kernel (3x3), don't forget the pooling layer
# Stick to rest of the settings!    
#


    # The fully conected part
    model.add(Flatten())
    model.add(Dense(128, activation='relu', kernel_initializer='he_uniform'))
    model.add(Dense(1, activation='sigmoid'))
    
    # compile model
    opt = SGD(learning_rate=0.001, momentum=0.9)
    model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
 
    return model
```

```python
# plot diagnostic learning curves
def summarize_diagnostics(history):
    
    pyplot.figure(figsize=(12, 4))

    # plot loss
    pyplot.subplot(2, 1, 1)
    pyplot.title('Cross Entropy Loss')
    pyplot.plot(history.history['loss'], color='blue', label='train')
    pyplot.plot(history.history['val_loss'], color='orange', label='validation')
    pyplot.ylim(0.0,1.0)
    pyplot.legend()
    
    # plot accuracy
    pyplot.subplot(2,1,2)
    pyplot.title('Classification Accuracy')
    pyplot.plot(history.history['accuracy'], color='blue', label='train')
    pyplot.plot(history.history['val_accuracy'], color='orange', label='validation')
    pyplot.legend()
    pyplot.ylim(0.0,1.0)
    
    pyplot.tight_layout()
    pyplot.show()
    
    # save plot to file
    #filename = sys.argv[0].split('/')[-1]
    #pyplot.savefig(filename + '_plot_CNN_8.png')
    #pyplot.close()
```

```python
# defining the model
model = define_model()
    
# create data generator
# scale image intensities between 0 and 1
datagen = ImageDataGenerator(rescale=1.0/255.0)

# prepare iterator
train_it = datagen.flow_from_directory('data-dogs-vs-cats/train/',
        class_mode='binary', batch_size=32, target_size=(200, 200))
valid_it = datagen.flow_from_directory('data-dogs-vs-cats/validation/',
        class_mode='binary', batch_size=32, target_size=(200, 200))
```

```python
# fit model -- this will take a while
history = model.fit(train_it, steps_per_epoch=len(train_it),
        validation_data=valid_it, validation_steps=len(valid_it), epochs=15, verbose=1)

## Task: Explain the verbose output!!
##
```

```python
# evaluate model
_, acc = model.evaluate(valid_it, steps=len(valid_it), verbose=1)

print('> %.3f' % (acc * 100.0))
    
```
